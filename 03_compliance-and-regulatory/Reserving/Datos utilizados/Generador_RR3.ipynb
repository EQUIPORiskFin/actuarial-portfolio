```python

# -*- coding: utf-8 -*-
"""
Ejemplo hipotético de base de datos para RR-4 (Daños) conforme al
Manual de Datos de Daños (CNSF, 2020-01-14).

- Dataset SINTÉTICO: sustituir claves/catálogos por las oficiales.
- PEA = PE * 365 / Duración original (en días).
- TipoRegistro "0": pólizas vigentes al corte.
- TipoRegistro "1": prima emitida directa SIN anualizar del año (1 registro por año).
- Duraciones ≤ 365/366 días.

Genera:
- base_rr4_danos_ejemplo.csv
- base_rr4_danos_ejemplo.xlsx
"""

import pandas as pd

def calcular_pea(prima_emitida_directa, duracion_promedio_dias):
    """PEA = PE * 365 / duración original; usar 366 si el corte es bisiesto."""
    if duracion_promedio_dias <= 0:
        raise ValueError("La duración promedio debe ser positiva.")
    return round(prima_emitida_directa * 365.0 / float(duracion_promedio_dias), 2)

# Placeholders (reemplazar por catálogos oficiales)
RAMOS = {"AUI":"Automóviles Individual","AUF":"Automóviles Flotilla","INC":"Incendio",
         "DMI":"Diversos Misceláneos","RCV":"Responsabilidad Civil","CAU":"Caución"}
TIPO_VEHICULO = {"AUI":"SEDAN","AUF":"CAMIONETA_FLOT","INC":"0","DMI":"0","RCV":"0","CAU":"0"}
COBERTURA = {"AUI":"RC_LIMITADA","AUF":"DAÑOS_MATERIALES","INC":"0","DMI":"0","RCV":"GENERAL","CAU":"GARANTIA"}
MONEDA_NACIONAL, MONEDA_EXTRANJERA = "10", "20"

registros = []
anios_corte = [2025, 2024]

# TipoRegistro "0": pólizas vigentes (valores ilustrativos)
for anio in anios_corte:
    # AUI
    pea = calcular_pea(3_000_000.00, 365)
    registros.append({
        "Año": anio, "RamoSubramo": "AUI", "TipoVehiculo": TIPO_VEHICULO["AUI"],
        "TipoRegistro": "0", "Cobertura": COBERTURA["AUI"], "Moneda": MONEDA_NACIONAL,
        "ClaveEsquemaReaseguro1": 10, "ClaveEsquemaReaseguro2": 20, "ClaveEsquemaReaseguro3": 0,
        "NumeroAsegurados": 1200, "PrimaEmitidaDirectaAnualizada": pea,
        "DuracionRemanentePromedio_dias": 180, "LimiteMaximoResponsabilidad": 50_000_000.00,
        "DuracionPromedio_dias": 365,
    })
    # AUF
    pea = calcular_pea(5_000_000.00, 300)
    registros.append({
        "Año": anio, "RamoSubramo": "AUF", "TipoVehiculo": TIPO_VEHICULO["AUF"],
        "TipoRegistro": "0", "Cobertura": COBERTURA["AUF"], "Moneda": MONEDA_NACIONAL,
        "ClaveEsquemaReaseguro1": 10, "ClaveEsquemaReaseguro2": 0, "ClaveEsquemaReaseguro3": 0,
        "NumeroAsegurados": 300, "PrimaEmitidaDirectaAnualizada": pea,
        "DuracionRemanentePromedio_dias": 150, "LimiteMaximoResponsabilidad": 120_000_000.00,
        "DuracionPromedio_dias": 300,
    })
    # INC
    pea = calcular_pea(8_500_000.00, 365)
    registros.append({
        "Año": anio, "RamoSubramo": "INC", "TipoVehiculo": TIPO_VEHICULO["INC"],
        "TipoRegistro": "0", "Cobertura": COBERTURA["INC"], "Moneda": MONEDA_EXTRANJERA,
        "ClaveEsquemaReaseguro1": 0, "ClaveEsquemaReaseguro2": 0, "ClaveEsquemaReaseguro3": 0,
        "NumeroAsegurados": 75, "PrimaEmitidaDirectaAnualizada": pea,
        "DuracionRemanentePromedio_dias": 200, "LimiteMaximoResponsabilidad": 300_000_000.00,
        "DuracionPromedio_dias": 365,
    })
    # DMI
    pea = calcular_pea(1_250_000.00, 180)
    registros.append({
        "Año": anio, "RamoSubramo": "DMI", "TipoVehiculo": TIPO_VEHICULO["DMI"],
        "TipoRegistro": "0", "Cobertura": COBERTURA["DMI"], "Moneda": MONEDA_NACIONAL,
        "ClaveEsquemaReaseguro1": 0, "ClaveEsquemaReaseguro2": 0, "ClaveEsquemaReaseguro3": 0,
        "NumeroAsegurados": 950, "PrimaEmitidaDirectaAnualizada": pea,
        "DuracionRemanentePromedio_dias": 60, "LimiteMaximoResponsabilidad": 45_000_000.00,
        "DuracionPromedio_dias": 180,
    })
    # RCV
    pea = calcular_pea(3_600_000.00, 365)
    registros.append({
        "Año": anio, "RamoSubramo": "RCV", "TipoVehiculo": TIPO_VEHICULO["RCV"],
        "TipoRegistro": "0", "Cobertura": COBERTURA["RCV"], "Moneda": MONEDA_NACIONAL,
        "ClaveEsquemaReaseguro1": 10, "ClaveEsquemaReaseguro2": 20, "ClaveEsquemaReaseguro3": 0,
        "NumeroAsegurados": 410, "PrimaEmitidaDirectaAnualizada": pea,
        "DuracionRemanentePromedio_dias": 220, "LimiteMaximoResponsabilidad": 500_000_000.00,
        "DuracionPromedio_dias": 365,
    })
    # CAU
    pea = calcular_pea(900_000.00, 270)
    registros.append({
        "Año": anio, "RamoSubramo": "CAU", "TipoVehiculo": TIPO_VEHICULO["CAU"],
        "TipoRegistro": "0", "Cobertura": COBERTURA["CAU"], "Moneda": MONEDA_NACIONAL,
        "ClaveEsquemaReaseguro1": 0, "ClaveEsquemaReaseguro2": 0, "ClaveEsquemaReaseguro3": 0,
        "NumeroAsegurados": 210, "PrimaEmitidaDirectaAnualizada": pea,
        "DuracionRemanentePromedio_dias": 120, "LimiteMaximoResponsabilidad": 60_000_000.00,
        "DuracionPromedio_dias": 270,
    })

# TipoRegistro "1": único registro por año con la prima del año (SIN anualizar)
registros += [
    {"Año": 2025, "RamoSubramo": "AUI", "TipoVehiculo": "0", "TipoRegistro": "1", "Cobertura": "0",
     "Moneda": MONEDA_NACIONAL, "ClaveEsquemaReaseguro1": 0, "ClaveEsquemaReaseguro2": 0,
     "ClaveEsquemaReaseguro3": 0, "NumeroAsegurados": 1,
     "PrimaEmitidaDirectaAnualizada": 4_500_000.00, "DuracionRemanentePromedio_dias": 1,
     "LimiteMaximoResponsabilidad": 0.00, "DuracionPromedio_dias": 365},
    {"Año": 2024, "RamoSubramo": "AUI", "TipoVehiculo": "0", "TipoRegistro": "1", "Cobertura": "0",
     "Moneda": MONEDA_NACIONAL, "ClaveEsquemaReaseguro1": 0, "ClaveEsquemaReaseguro2": 0,
     "ClaveEsquemaReaseguro3": 0, "NumeroAsegurados": 1,
     "PrimaEmitidaDirectaAnualizada": 4_000_000.00, "DuracionRemanentePromedio_dias": 1,
     "LimiteMaximoResponsabilidad": 0.00, "DuracionPromedio_dias": 365},
]

cols = ["Año","RamoSubramo","TipoVehiculo","TipoRegistro","Cobertura","Moneda",
        "ClaveEsquemaReaseguro1","ClaveEsquemaReaseguro2","ClaveEsquemaReaseguro3",
        "NumeroAsegurados","PrimaEmitidaDirectaAnualizada",
        "DuracionRemanentePromedio_dias","LimiteMaximoResponsabilidad",
        "DuracionPromedio_dias"]

df = pd.DataFrame(registros, columns=cols)

# Guardar archivos
df.to_csv("base_rr4_danos_ejemplo.csv", index=False, encoding="utf-8")
df.to_excel("base_rr4_danos_ejemplo.xlsx", index=False, engine="openpyxl")
```

```python


# -*- coding: utf-8 -*-
"""
Generador de base ficticia RR-4 (Daños) con 30,000 registros
conforme al Manual de Datos de Daños (CNSF, 2020-01-14).

Archivos:
- rr4_danos_ficticio_30000.csv
- rr4_danos_ficticio_30000.xlsx
"""

import random
import pandas as pd

random.seed(12345)

SUBRAMOS = ["AUI", "AUF", "INC", "DMI", "RCV", "CAU"]
AUTO_TIPOS = ["SEDAN", "SUV", "PICKUP", "CAMIONETA_FLOT"]
AUTO_COBERTURAS = ["RC_LIMITADA", "RC_AMPLIA", "DAÑOS_MATERIALES", "ROBO"]
NO_APLICA = "0"
MONEDA_NAL, MONEDA_EXT = "10", "20"
ANIOS = list(range(2021, 2026))

def es_bisiesto(anio):
    return (anio % 4 == 0 and (anio % 100 != 0 or anio % 400 == 0))

TOTAL_REGISTROS = 30000
TR1_REGISTROS = len(ANIOS) * len(SUBRAMOS)  # 30
TR0_REGISTROS = TOTAL_REGISTROS - TR1_REGISTROS  # 29970
por_combo = TR0_REGISTROS // (len(ANIOS) * len(SUBRAMOS))  # 999
assert TR0_REGISTROS == por_combo * len(ANIOS) * len(SUBRAMOS)

def generar_tr0(anio, subramo):
    dur_max = 366 if es_bisiesto(anio) else 365
    duracion_promedio = random.randint(60, dur_max)
    duracion_remanente = random.randint(1, duracion_promedio)

    # Prima emitida directa (PE) sintética (lognormal)
    pe = random.lognormvariate(mu=13.0, sigma=0.6)
    pe = min(pe, 50_000_000.0)
    pea = round(pe * 365.0 / float(duracion_promedio), 2)

    if subramo in ("AUI", "AUF"):
        num_asegurados = random.randint(1, 5000)
    elif subramo == "RCV":
        num_asegurados = random.randint(1, 2000)
    elif subramo == "INC":
        num_asegurados = random.randint(1, 500)
    elif subramo == "DMI":
        num_asegurados = random.randint(1, 3000)
    elif subramo == "CAU":
        num_asegurados = random.randint(1, 1000)
    else:
        num_asegurados = random.randint(1, 1000)

    moneda = MONEDA_NAL if random.random() < 0.85 else MONEDA_EXT

    if random.random() < 0.6:
        rea1 = 10
        rea2 = 20 if random.random() < 0.4 else 0
        rea3 = 0
    else:
        rea1 = rea2 = rea3 = 0

    if subramo in ("INC", "RCV"):
        lmr = round(random.uniform(50_000_000, 800_000_000), 2)
    elif subramo in ("AUI", "AUF"):
        lmr = round(random.uniform(5_000_000, 200_000_000), 2)
    else:
        lmr = round(random.uniform(10_000_000, 150_000_000), 2)

    if subramo in ("AUI", "AUF"):
        tipo_vehiculo = random.choice(AUTO_TIPOS)
        cobertura = random.choice(AUTO_COBERTURAS)
    else:
        tipo_vehiculo = NO_APLICA
        cobertura = NO_APLICA

    return {
        "Año": anio,
        "RamoSubramo": subramo,
        "TipoVehiculo": tipo_vehiculo,
        "TipoRegistro": "0",
        "Cobertura": cobertura,
        "Moneda": moneda,
        "ClaveEsquemaReaseguro1": rea1,
        "ClaveEsquemaReaseguro2": rea2,
        "ClaveEsquemaReaseguro3": rea3,
        "NumeroAsegurados": num_asegurados,
        "PrimaEmitidaDirectaAnualizada": pea,
        "DuracionRemanentePromedio_dias": duracion_remanente,
        "LimiteMaximoResponsabilidad": lmr,
        "DuracionPromedio_dias": duracion_promedio,
    }

registros = []
for anio in ANIOS:
    for subramo in SUBRAMOS:
        for _ in range(por_combo):
            registros.append(generar_tr0(anio, subramo))

for anio in ANIOS:
    for subramo in SUBRAMOS:
        prima_anual = round(random.uniform(5_000_000, 150_000_000), 2)
        registros.append({
            "Año": anio,
            "RamoSubramo": subramo,
            "TipoVehiculo": NO_APLICA,
            "TipoRegistro": "1",
            "Cobertura": NO_APLICA,
            "Moneda": MONEDA_NAL,
            "ClaveEsquemaReaseguro1": 0,
            "ClaveEsquemaReaseguro2": 0,
            "ClaveEsquemaReaseguro3": 0,
            "NumeroAsegurados": 1,
            "PrimaEmitidaDirectaAnualizada": prima_anual,  # TR1: prima del año SIN anualizar
            "DuracionRemanentePromedio_dias": 1,
            "LimiteMaximoResponsabilidad": 0.0,
            "DuracionPromedio_dias": 365,
        })

cols = [
    "Año","RamoSubramo","TipoVehiculo","TipoRegistro","Cobertura","Moneda",
    "ClaveEsquemaReaseguro1","ClaveEsquemaReaseguro2","ClaveEsquemaReaseguro3",
    "NumeroAsegurados","PrimaEmitidaDirectaAnualizada",
    "DuracionRemanentePromedio_dias","LimiteMaximoResponsabilidad","DuracionPromedio_dias"
]
df = pd.DataFrame(registros, columns=cols)

assert len(df) == TOTAL_REGISTROS
assert (df["DuracionPromedio_dias"] <= 366).all()
assert (df["DuracionRemanentePromedio_dias"] <= 366).all()
mask_rea0 = df["ClaveEsquemaReaseguro1"] == 0
assert (df.loc[mask_rea0, ["ClaveEsquemaReaseguro2","ClaveEsquemaReaseguro3"]] == 0).all().all()

df.to_csv("rr4_danos_ficticio_30000.csv", index=False, encoding="utf-8")
df.to_excel("rr4_danos_ficticio_30000.xlsx", index=False, engine="openpyxl")
```
